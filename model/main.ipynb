# Movie Recommendation System

### Loading Libraries & Dataset

```python
import pandas as pd
```

```python
data = pd.read_csv('assets/movies.csv')
```

```python
data.info()
data.columns
```

```python
data.head()
```

### Trimming down to important columns

```python
data = data[
    [
        'id', 
        'title',
        'release_date',
        'rating',
        'description',
        'writers', 
        'directors', 
        'stars', 
        'genres' 
    ]
]
data.head()
```

### Dropping null values

```python
data = data[data['description'].notna()]
data = data[data['writers'].notna()]
data = data[data['stars'].notna()]
data = data[data['genres'].notna()]
data = data[data['directors'].notna()]
data.info()
```

```python
data["rating"] = data["rating"].fillna(0)
data.head()
```

### Tag prepration

```python
def to_list(x):
    import ast
    return ast.literal_eval(x)[:7]

def description_to_list(x):
    return [i for i in x.split(' ')]

def title_to_list(x):
    return [i for i in x.split(' ')]

def limiter(x):
    return x[:7]
```

```python
# List Conversion
data['writers'] = data['writers'].apply(to_list)
data['directors'] = data['directors'].apply(to_list)
data['stars'] = data['stars'].apply(to_list)
data['genres'] = data['genres'].apply(to_list)

data['description'] = data['description'].apply(description_to_list)
data['title_for_tag'] = data['title'].apply(title_to_list)

data.head()
```

```python
data['tag'] = data['title_for_tag'] + data['description'] + data['writers'] + data['directors'] + data['stars'] + data['genres']
data["tag"] = data["tag"].apply(lambda x: " ".join(x).lower())
data = data[["id", "title","release_date", "rating", "tag"]]
data.head()
```

### Stemming the tag

```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
def stem(text):
    y = [ps.stem(i) for i in text.split()]
    return " ".join(y)
```

```python
data['tag'] = data['tag'].apply(stem)
```

```python
data.head()
```

### Vectorizing with TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features=15000, stop_words='english', ngram_range=(1,2))
vector = tfidf.fit_transform(data['tag'])
```

### Setting up Near Neighbour

```python
from sklearn.neighbors import NearestNeighbors
model = NearestNeighbors(n_neighbors=6, metric='cosine', algorithm='brute')
model.fit(vector)
```

```python
position = data[data["title"] == "Transformers: Age of Extinction"]

if position.empty:
    print("Position not found")
else:
    iloc_index = data.index.get_loc(position.index[0])
    iloc_index
```

```python
_, indices = model.kneighbors(vector[iloc_index], n_neighbors=6)
for i in indices[0][1:]:
    print(data.iloc[i]["title"])
```

### Saving the Dataset, Vector and Model

Run the below code to save the dataset, vector and model\
It can then be used to further consumend as API

```python
import pickle

with open('assets/vector.pkl', 'wb') as f:
    pickle.dump(vector, f)

with open('assets/model.pkl', 'wb') as f:
    pickle.dump(model, f)

with open('assets/data.pkl', 'wb') as f:
    pickle.dump(data, f)
```
